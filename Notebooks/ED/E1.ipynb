## Berechnung von Rx1, Rx2, Rx3

```python
# mit den Original Matrizen
R123_mat = R_A1_arr * L_A1_arr / (1-L_A1_arr)
R_mat = R_A1_arr
L_mat = L_A1_arr
```

```python
def R123_err(R, L, u_R, u_L, l=np.ones(8)):
    t1 = (L / (1-L))**2 * u_R**2
    t2 = (R * (1/(1-L) + L/(1-L)**2))**2 * u_L**2
    return np.round(np.sqrt(t1 + t2), 4)
```
